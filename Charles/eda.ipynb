```python
import os
import json
import pandas as pd
from dotenv import load_dotenv

load_dotenv()

DATASET_PATH: str = os.getenv("DATASET_PATH", "")
os.chdir(DATASET_PATH)
```

```python
with open('train.json', 'r', encoding='utf-8') as f:
	train_data = json.load(f)
print("Example train data entry:", train_data["7"]['Overall_class'])
```

```python
ax_t1_radiomics_train = pd.read_csv('radiomics_info\\train\\ax_t1_radiomics_train.csv').rename(columns={
    'rad_firstorder_Mean': 'ax_t1_rad_firstorder_Mean', 
    'rad_firstorder_Entropy': 'ax_t1_rad_firstorder_Entropy', 
    'rad_firstorder_90Percentile': 'ax_t1_rad_firstorder_90Percentile', 
    'rad_glcm_Contrast': 'ax_t1_rad_glcm_Contrast', 
    'rad_glcm_JointEntropy': 'ax_t1_rad_glcm_JointEntropy'
})
ax_t1_radiomics_train.drop(['sex', 'age', 'modality'], axis=1, inplace=True)


ax_t1c_radiomics_train = pd.read_csv('radiomics_info\\train\\ax_t1c_radiomics_train.csv').rename(columns={
    'rad_firstorder_Mean': 'ax_t1c_rad_firstorder_Mean', 
    'rad_firstorder_Entropy': 'ax_t1c_rad_firstorder_Entropy', 
    'rad_firstorder_90Percentile': 'ax_t1c_rad_firstorder_90Percentile', 
    'rad_glcm_Contrast': 'ax_t1c_rad_glcm_Contrast', 
    'rad_glcm_JointEntropy': 'ax_t1c_rad_glcm_JointEntropy'
})
ax_t1c_radiomics_train.drop(['sex', 'age', 'modality'], axis=1, inplace=True)


ax_t2_radiomics_train = pd.read_csv('radiomics_info\\train\\ax_t2_radiomics_train.csv').rename(columns={
    'rad_firstorder_Mean': 'ax_t2_rad_firstorder_Mean', 
    'rad_firstorder_Entropy': 'ax_t2_rad_firstorder_Entropy', 
    'rad_firstorder_90Percentile': 'ax_t2_rad_firstorder_90Percentile', 
    'rad_glcm_Contrast': 'ax_t2_rad_glcm_Contrast', 
    'rad_glcm_JointEntropy': 'ax_t2_rad_glcm_JointEntropy'
})
ax_t2_radiomics_train.drop(['sex', 'age', 'modality'], axis=1, inplace=True)

ax_t2f_radiomics_train = pd.read_csv('radiomics_info\\train\\ax_t2f_radiomics_train.csv').rename(columns={
    'rad_firstorder_Mean': 'ax_t2f_rad_firstorder_Mean', 
    'rad_firstorder_Entropy': 'ax_t2f_rad_firstorder_Entropy', 
    'rad_firstorder_90Percentile': 'ax_t2f_rad_firstorder_90Percentile', 
    'rad_glcm_Contrast': 'ax_t2f_rad_glcm_Contrast', 
    'rad_glcm_JointEntropy': 'ax_t2f_rad_glcm_JointEntropy'
})
ax_t2f_radiomics_train.drop(['sex', 'age', 'modality'], axis=1, inplace=True)
```

```python
merged_inner = ax_t1_radiomics_train.merge(ax_t1c_radiomics_train, on='case_id') \
                  .merge(ax_t2_radiomics_train, on='case_id') \
                  .merge(ax_t2f_radiomics_train, on='case_id')
merged_inner
```

```python
type_mapping = {int(case_id): info['Overall_class'] for case_id, info in train_data.items()}

merged_inner["Type"] = merged_inner["case_id"].map(type_mapping)

# check
merged_inner["Type"].value_counts(dropna=False)
```

```python
merged_inner_filtered = merged_inner.dropna()
merged_inner_filtered
```

```python

merged_inner_filtered["Type"] = merged_inner_filtered["case_id"].map(type_mapping)

# check
merged_inner_filtered["Type"].value_counts(dropna=False)
```

```python
import seaborn as sns

import matplotlib.pyplot as plt

numeric_features = [c for c in merged_inner_filtered.select_dtypes(include="number").columns if c != "case_id"]

df_long = merged_inner_filtered.melt(
	id_vars="Type",
	value_vars=numeric_features,
	var_name="feature",
	value_name="value"
)

plt.figure(figsize=(18, 10))
sns.boxplot(data=df_long, x="feature", y="value", hue="Type")
plt.xticks(rotation=90)
plt.title("Feature distributions by Type (merged_inner_filtered)")
plt.tight_layout()
plt.show()
```

```python
import sklearn.linear_model

X = merged_inner_filtered[merged_inner_filtered.columns.difference(['Type'])]
y = merged_inner_filtered['Type']
```

# Logistic Regression: Liblinear

```python
from sklearn.metrics import accuracy_score, f1_score
from sklearn.preprocessing import StandardScaler, RobustScaler

X_train, X_test, y_train, y_test = sklearn.model_selection.train_test_split(X, y, test_size=0.3, stratify=y)
# Fit scaler ONLY on training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Learns mean/std from train

# Transform both train and test using TRAIN parameters
X_test_scaled = scaler.transform(X_test)  # Uses train's mean/std


model = sklearn.linear_model.LogisticRegression(solver='liblinear', penalty='l1', class_weight='balanced')
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate scores
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, average='micro')

# Output both metrics
print(f"Accuracy: {accuracy}")
print(f"F1 Score: {f1}")
```

# SMOTE

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline as ImbPipeline

X_train, X_test, y_train, y_test = sklearn.model_selection.train_test_split(X, y, test_size=0.3, stratify=y)
# Fit scaler ONLY on training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Learns mean/std from train

# Transform both train and test using TRAIN parameters
X_test_scaled = scaler.transform(X_test)  # Uses train's mean/std
print("Before SMOTE:", pd.Series(y_train).value_counts())


smote = SMOTE(random_state=42, k_neighbors=3)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_scaled, y_train)

print("After SMOTE:", pd.Series(y_train_resampled).value_counts())

model = sklearn.linear_model.LogisticRegression(solver='liblinear', penalty='l1', class_weight='balanced')
model.fit(X_train_resampled, y_train_resampled)

# Make predictions
y_pred = model.predict(X_test)

# Calculate scores
accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred, average='micro')

# Output both metrics
print(f"Accuracy: {accuracy}")
print(f"F1 Score: {f1}")
```

# Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, f1_score, classification_report

X_train, X_test, y_train, y_test = sklearn.model_selection.train_test_split(X, y, test_size=0.3, stratify=y)
# Fit scaler ONLY on training data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Learns mean/std from train

# Transform both train and test using TRAIN parameters
X_test_scaled = scaler.transform(X_test)  # Uses train's mean/std

smote = SMOTE(random_state=42, k_neighbors=3)
X_train_resampled, y_train_resampled = smote.fit_resample(X_train_scaled, y_train)

print("After SMOTE:", pd.Series(y_train_resampled).value_counts())

model = RandomForestClassifier(
    n_estimators=200,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

model.fit(X_train_resampled, y_train_resampled)
y_pred = model.predict(X_test_scaled)

print(f"Accuracy:    {accuracy_score(y_test, y_pred):.4f}")
print(f"Macro F1:    {f1_score(y_test, y_pred, average='macro'):.4f}")
print("\nPer-class breakdown:")
print(classification_report(y_test, y_pred))
```
